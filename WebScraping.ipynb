Scraping the countries data from https://www.scrapethissite.com/pages/simple/

```python
import requests
```

# Common Get requests statuses
200 => All Okay
301/302 => Redirect
400 => Bad Request
401 => Unauthorized
403 => Forbidden
404 => Server not found
500 => Server error

```python
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
if res.status_code == 200:
    # print(res.text) # All the HTML of the page as a string
    # print(res.content) # Raw data
    print(res.headers) # Meta data

with open("scraped_data/ScrapedData1.html","w") as f:
    f.write(res.text)
```
